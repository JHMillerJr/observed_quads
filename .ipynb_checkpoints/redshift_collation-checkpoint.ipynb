```python
#> observed lens file
obsFile = '../observed_quads/image_positions/all_quads_positions.csv'
df = pd.read_csv(obsFile)

#> collecting redshifts
redshifts = []
for name in df['Name'].unique():
    
    #> getting redshifts
    galaxy = df[ df['Name'] == name ]
    zs = galaxy['zsource'].unique()
    zl = galaxy['zdef'].unique()
    
    #> error handling
    if (len(zs) != 1) or (len(zl) != 1): error.phrase(f'Lens {name} contains either multiple zl: {zl} or zs: {zs}')
    
    #> saving
    redshifts.append([name, zl[0], zs[0]])

redshifts_df = pd.DataFrame(redshifts, columns=['Name', 'zl', 'zs'])
redshifts_df.to_csv('../observed_quads/redshifts.csv', index=False)
```
